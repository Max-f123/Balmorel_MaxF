To sum-up:

# 3) Installed Capacity

Lower DR leads to more installed capacity in the power system, for generation and transport

### Arguments:

RES have a lower NCF:

And more intermitency:

So need more production capacity, transmission and storage in a park that contains more RES, to meet the demand that remains unchanged with different DF

To sum up:
- VRE requires oversizing the system to handle intermittency(production profiles varying over time) and meet demand reliably (lower NCF)
- Higher DR leads to preference for flexible, dispatchable plants (e.g., gas), reducing the need for system overcapacity.
- Therefore, total system capacity tends to be lower under high DR, and higher under low DR.
